```python
month_order = [
    'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
    'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'
]

df['Month'] = pd.Categorical(
    df['Month'],
    categories=month_order,
    ordered=True
)
```

```python
monthly_cost = (
    df.groupby('Month', observed=True)['Monthly_Cost_INR']
      .sum()
)

monthly_cost
```

```python
plt.figure(figsize=(12, 6))
monthly_cost.plot(kind='line', marker='o',color='purple')
plt.title('Monthly Cloud Cost Trend')
plt.xlabel('Month')
plt.ylabel('Cost (INR)')
plt.xticks(range(12), [ 'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun','Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'])
plt.show()
```

```python
dept_analysis = df.groupby('Department').agg(Total_Cost=('Monthly_Cost_INR','sum'),Potential_Savings=('Potential_Saving_INR','sum'),
                                            Resource_count=('Resource_Id','count'))
dept_analysis.sort_values('Potential_Savings',ascending=False)
```

```python
current_cost = df['Monthly_Cost_INR'].sum()
potential_savings = df['Potential_Saving_INR'].sum()
optimized_cost = current_cost - potential_savings
```

```python
print('Current Cost:',round(current_cost,2))
print('Potential_Savings:',round(potential_savings,2))
print('Optimized Cost:',round(optimized_cost,2))
```

```python
savings_pct = (potential_savings / current_cost) * 100
print('Savings Percentage:',round(savings_pct,3))
```

```python
summary = {
    'Total Cloud Cost': round(float(current_cost),2),
    'Potential Savings': round(float(potential_savings),2),
    'Optimized Cost': round(float(optimized_cost),2),
    'Savings Percentage': round(float(savings_pct),2),
    'Underutilized Resources': len(underutilized)
}

summary
```
